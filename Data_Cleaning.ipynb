```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df['Sale_Date'] = pd.to_datetime(df['Sale_Date'])
```

```python
df.dtypes
```

```python
#Check invalid/negative values
(df['Quantity_Sold'] < 0).sum()
(df['Sales_Amount'] < 0).sum()
(df['Unit_Cost'] < 0).sum()
(df['Unit_Price'] < 0).sum()
```

```python
df[['Sales_Amount', 'Quantity_Sold', 'Unit_Price']].head()
#for calculating revenue
```

```python
#Calculate estimated profit
df['Total_Cost'] = df['Unit_Cost'] * df['Quantity_Sold']
df['Profit'] = df['Sales_Amount'] - df['Total_Cost']
df[['Sales_Amount', 'Total_Cost', 'Profit']].head()
```

```python
#Calculate profit margin
df['Profit_Margin'] = (df['Profit'] / df['Sales_Amount']) * 100
df[['Sales_Amount', 'Profit', 'Profit_Margin']].head()
```

```python
#Create Year Column
df['Year'] = df['Sale_Date'].dt.year
#create month column
df['Month'] = df['Sale_Date'].dt.month
#create month name
df['Month_Name'] = df['Sale_Date'].dt.month_name()
#create Quarter
df['Quarter'] = df['Sale_Date'].dt.quarter
df[['Sale_Date', 'Year', 'Month', 'Month_Name', 'Quarter']].head()
```
