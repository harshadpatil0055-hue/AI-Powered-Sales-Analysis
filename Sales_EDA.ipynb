```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_sales_data.csv")
```

```python
df['Sale_Date'] = pd.to_datetime(df['Sale_Date']) #converting datatypes
```

```python
#total sales
total_sales = df['Sales_Amount'].sum()
print("Total Sales:", total_sales)
```

```python
#total cost
total_cost = df['Total_Cost'].sum()
print("Total Cost:", total_cost)
```

```python
#total profit
total_profit = df['Profit'].sum()
print("Total Profit:", total_profit)
```

```python
#Average Sales
average_sales = df['Sales_Amount'].mean()
print("Average Sales per Order:", average_sales)
```

```python
#total Quantity Sold
total_quantity = df['Quantity_Sold'].sum()
print("Total Quantity Sold:", total_quantity)
```

```python
#Sales by region
region_sales = df.groupby('Region')['Sales_Amount'].sum().sort_values(ascending=False)
region_sales
```

```python
#sales by region plot
region_sales.plot(kind='bar', figsize=(8,5))
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
#Profit By Region
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
region_profit
```

```python
#profit by region plot
region_profit.plot(kind='bar', figsize=(8,5))
plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()
```

```python
#sales by product category
category_sales = df.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending=False)
category_sales
```

```python
#visualization
category_sales.plot(kind='bar', figsize=(8,5))
plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
#profit by product category
category_profit = df.groupby('Product_Category')['Profit'].sum().sort_values(ascending=False)
category_profit
```

```python
#visualizations
category_profit.plot(kind='bar', figsize=(8,5))
plt.title("Profit by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
#top sales representatives
rep_sales = df.groupby('Sales_Rep')['Sales_Amount'].sum().sort_values(ascending=False)
rep_sales
```

```python
#top10 repeat sales
top_10_reps = rep_sales.head(10)
top_10_reps
```

```python
#visualizations
top_10_reps.plot(kind='bar', figsize=(10,5))
plt.title("Top 10 Sales Representatives")
plt.xlabel("Sales Representative")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
#monthly sales analysis
monthly_sales = df.groupby('Month')['Sales_Amount'].sum()
monthly_sales
```

```python
#visualizations
monthly_sales.plot(kind='line', marker='o', figsize=(10,5))
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.grid()
plt.show()
```

```python
# monthly profit
monthly_profit = df.groupby('Month')['Profit'].sum()
monthly_profit
```

```python
#visualization
monthly_profit.plot(kind='line', marker='o', figsize=(10,5))
plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.grid()
plt.show()
```

```python
#sales by customer type
customer_type_sales = df.groupby('Customer_Type')['Sales_Amount'].sum().sort_values(ascending=False)
customer_type_sales
```

```python
#visualization
customer_type_sales.plot(kind='bar', figsize=(8,5))
plt.title("Sales by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Sales Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
#sales by sales channel
channel_sales = df.groupby('Sales_Channel')['Sales_Amount'].sum().sort_values(ascending=False)
channel_sales
```

```python
#visualization
channel_sales.plot(kind='bar', figsize=(8,5))
plt.title("Sales by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Sales Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
#payment method analysis
payment_sales = df.groupby('Payment_Method')['Sales_Amount'].sum().sort_values(ascending=False)
payment_sales
```

```python
#visualization
payment_sales.plot(kind='line',marker="o", figsize=(8,5))
plt.title("Sales by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
#top products
product_sales = df.groupby('Product_ID')['Sales_Amount'].sum().sort_values(ascending=False)
product_sales.head(10)
```

```python
#visualization
top_products = product_sales.head(10)
top_products.plot(kind='bar', figsize=(10,5))
plt.title("Top 10 Products by Sales")
plt.xlabel("Product ID")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
#discount analysis\
print("Average Discount:", df['Discount'].mean())
#max discount
print("Maximum Discount:", df['Discount'].max())
#min discount
print("Minimum Discount:", df['Discount'].min())
```

```python
#visualization discount vs profit
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Discount',
    y='Profit'
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```

```python
#visualization Sales vs Quantity
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Quantity_Sold',
    y='Sales_Amount'
)

plt.title("Quantity Sold vs Sales")
plt.xlabel("Quantity Sold")
plt.ylabel("Sales Amount")
plt.show()
```

```python
#profit margin analysis
print("Average Profit Margin:",
      df['Profit_Margin'].mean())
#profit margin by category
category_margin = df.groupby('Product_Category')['Profit_Margin'].mean().sort_values(ascending=False)

category_margin
```

```python
#visualizations
category_margin.plot(kind='bar', figsize=(10,5))
plt.title("Average Profit Margin by Category")
plt.xlabel("Product Category")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
#correlation Analysis
numeric_columns = [
    'Sales_Amount',
    'Quantity_Sold',
    'Unit_Cost',
    'Unit_Price',
    'Discount',
    'Total_Cost',
    'Profit',
    'Profit_Margin'
]

correlation = df[numeric_columns].corr()

correlation
```

```python
#visualizations
plt.figure(figsize=(10,7))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
#find the best and worst sales months
best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()

worst_month = monthly_sales.idxmin()
worst_month_sales = monthly_sales.min()

print("Best Month:", best_month)
print("Best Month Sales:", best_month_sales)

print("Worst Month:", worst_month)
print("Worst Month Sales:", worst_month_sales)
```
